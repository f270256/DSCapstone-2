```python
# Convert launched/deadline to datetime

df["launched"] = pd.to_datetime(df["launched"], errors="coerce")
df["deadline"] = pd.to_datetime(df["deadline"], errors="coerce")
```

```python
df["launch_year"] = df["launched"].dt.year
df["launch_month"] = df["launched"].dt.month
df["launch_weekday"] = df["launched"].dt.weekday
```

```python
df = df.drop(columns=["launched", "deadline"])
```

```python
drop_cols = [
    "ID", "name", "state",
    "pledged", "usd_pledged_real",
    "backers"
]
```

```python
df = df.drop(columns=[c for c in drop_cols if c in df.columns])
```

```python
if "goal" in df.columns and "usd_goal_real" in df.columns:
    df = df.drop(columns=["goal"])
```

```python
X = df.drop("state_binary", axis=1)
y = df["state_binary"]
```

```python
X.dtypes
```

```python
categorical_features = ["category", "main_category", "currency", "country"]
categorical_features = [c for c in categorical_features if c in X.columns]

numeric_features = X.drop(columns=categorical_features).columns.tolist()
```

```python
print("Categorical features:", categorical_features)
print("Numeric features:", numeric_features)
```

```python
X_encoded = pd.get_dummies(X, columns=categorical_features, drop_first=True)
```

```python
X_encoded.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("Train shape:", X_train.shape, "Test shape:", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()
```

```python
X_train_scaled[numeric_features] = scaler.fit_transform(X_train[numeric_features])
X_test_scaled[numeric_features] = scaler.transform(X_test[numeric_features])
```
